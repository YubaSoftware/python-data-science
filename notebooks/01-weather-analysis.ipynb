# Exploring the sample weather data

This notebook reads the small synthetic weather CSV, computes a few summaries, and plots temperature over time. Run it from the repository root or the `notebooks/` directory. The chart cell requires Matplotlib (`py -m pip install -r requirements.txt`).

```python
import csv
from pathlib import Path
from statistics import mean, median

root = Path.cwd()
if not (root / "datasets").exists():
    root = root.parent

with (root / "datasets" / "weather.csv").open(encoding="utf-8", newline="") as file:
    weather = list(csv.DictReader(file))

temperatures = [float(row["temperature"]) for row in weather]
print(f"Observations: {len(weather)}")
print(f"Mean temperature: {mean(temperatures):.1f} C")
print(f"Median temperature: {median(temperatures):.1f} C")
```

The mean and median are close for this small example, but neither number shows when temperatures changed. A line chart makes the sequence visible.

```python
import matplotlib.pyplot as plt

dates = [row["date"] for row in weather]
fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(dates, temperatures, marker="o")
ax.set(title="Sample daily temperatures", xlabel="Date", ylabel="Temperature (C)")
ax.tick_params(axis="x", rotation=45)
fig.tight_layout()
plt.show()
```
